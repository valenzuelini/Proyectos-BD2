```python
from pathlib import Path
import pandas as pd

carpeta = Path("../data/processed")

tiempos = pd.concat(
    [pd.read_csv(f) for f in sorted(carpeta.glob("tiempos_*.csv"))],
    ignore_index=True,
)

print("bases encontradas:", tiempos["base"].unique().tolist())
print("mediciones por base y K:", tiempos.groupby(["base", "k"]).size().unique().tolist())

resumen = tiempos.groupby(["base", "k"])["ms"].agg(["min", "max", "mean"]).round(3)
resumen
```

```python
import matplotlib.pyplot as plt

promedios = resumen["mean"].unstack("base")

ax = promedios.plot(kind="bar", figsize=(7, 4), rot=0)
ax.set_title("Tiempo promedio de búsqueda por K")
ax.set_xlabel("K (resultados pedidos)")
ax.set_ylabel("Tiempo promedio (ms)")
plt.tight_layout()

plt.savefig("../evidencias/comparacion_tiempos.png", dpi=150)
plt.show()
```
